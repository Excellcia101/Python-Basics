```python
import numpy as np
```

```python
# Creating the new array
np.random.seed(42)
daily_sales = np.random.normal(loc=100, scale=30, size=(10,4))
print(daily_sales.shape)
# Reshaping the gaddamn arrays
reshaped = daily_sales.reshape(5,8)
print(reshaped.round(2))
transposed = daily_sales.T
print(transposed.round(2))
trans_mean = np.mean(transposed.round(2))
og_mean = np.mean(daily_sales.round(2))
print(f"The transposed mean: {trans_mean.round(0)}")
print(f"The Original Mean: {og_mean.round(0)}")
if og_mean == trans_mean:
    print("They are equal")
else:
    print("Not Equal")
```
