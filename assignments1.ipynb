```python
subgrid = daily_sales.iloc[0:5, [0, 2]]
print(subgrid)
print(np.shape(subgrid))

even = subgrid.iloc[::2]
print(even)

total = subgrid.sum()
print(f"Total for the 5 days{total.values}")

even_total = subgrid.iloc[::2].sum()
print(f"Total for the even number of days{even_total.values}")
```
