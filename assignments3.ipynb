```python
import numpy as np
```

```python
np.random.seed(7)
seeded_data = np.random.normal(loc=30000, scale= 10000, size= (10,4))
daily_sales = seeded_data.round(0)
counting = np.where(daily_sales.copy() > 20000, 1, 0)
co = list(counting)
print(counting)

```

```python
clothes = daily_sales[:,2]
premium = clothes >= 10000
standard = (clothes >= 7500) & (clothes < 10000)
below = clothes < 7500
print(clothes[premium])
print(clothes[standard])
print(clothes[below])
```

```python

electronics = daily_sales[:,1]
elec_copy = electronics.copy()
elec_mean = sum(electronics) / len(electronics)
print(np.where(elec_copy < 40000,elec_mean,elec_copy))
```

```python
top5_values = daily_sales[0:4,0:4]
rows, cols = np.where(top5_values)
print(top5_values[rows,cols])
```
